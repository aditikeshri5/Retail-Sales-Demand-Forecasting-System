```python
# Import required libraries

import pandas as pd
import numpy as np
import os

print("Libraries imported successfully!")
```

```python
# Load the original dataset

df = pd.read_csv("../data/raw/train.csv")

# Display first five records

df.head()
```

```python
# Check dataset dimensions

print("Dataset Shape:", df.shape)

# Check column names

print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and non-null values

print("\nDataset Information:")
df.info()
```

```python
# Convert date column into datetime format

df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Verify the conversion

print("Date Data Type:", df['date'].dtype)

print("\nStarting Date:", df['date'].min())
print("Ending Date:", df['date'].max())
```

```python
# Check missing values

print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records

print("\nDuplicate Records:", df.duplicated().sum())

# Check data types

print("\nData Types:")
print(df.dtypes)
```

```python
# Check for invalid dates
print("Invalid Dates:", df['date'].isnull().sum())

# Check negative sales values
print("Negative Sales:", (df['sales'] < 0).sum())

# Check invalid store IDs
print("Invalid Store IDs:",
      (~df['store'].between(1, 10)).sum())

# Check invalid item IDs
print("Invalid Item IDs:",
      (~df['item'].between(1, 50)).sum())

# Check sales statistics
print("\nSales Statistics:")
print(df['sales'].describe())
```

```python
# Sort data by date, store and item

df = df.sort_values(
    by=['date', 'store', 'item']
).reset_index(drop=True)

print("Dataset sorted successfully!")

df.head()
```

```python
# Create processed data directory if it doesn't exist

os.makedirs("../data/processed", exist_ok=True)

# Save cleaned dataset

df.to_csv(
    "../data/processed/cleaned_sales.csv",
    index=False
)

print("Preprocessed dataset saved successfully!")
print("Saved location: data/processed/cleaned_sales.csv")
```
